```python
from scipy import io
from matplotlib import pyplot as plt
```

```python
dc= io.loadmat(file_name='sampleECG2.mat')
dc.keys()
```

```python
sig= dc['signal']
plt.subplot(2,1,1)
plt.plot(sig)
plt.show()
```

```python
dc1={'signal':sig,'fs':720}
io.savemat(file_name='sampleECG5.mat',mdict=dc1)
```
